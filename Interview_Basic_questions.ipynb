**Q12. Write a program to find the Fibonacci series up to n.**

```python
def fibonacci_series(n)-> list:
    fibonacci =[0,1]
    while len(fibonacci) < n:
        fibonacci.append(fibonacci[-1]+fibonacci[-2])
    return fibonacci[:n]
n= int(input("Enter a number: "))
print(fibonacci_series(n))
```

**Q17. Write a Python function to find duplicate elements in a list.**

```python
def find_duplicates(lst)-> list:
    seen=set()
    duplicates=set()
    for item in lst:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)
    return list(duplicates)
n = [20,34,40,30,10,20,50,60,40,20,10]
print(sorted(find_duplicates(n)))
    
```

**Q18. Write a Python function to check if a number is prime.**

```python
def is_prime(num)->str:
    if num <=1:
        return "Not a Prime Number !!"
    for i in range(2,int(num**0.5)+1):
        if num%i==0:
            return "Not a Prime Number!!"
    return "Prime Number!!"
print(is_prime(457))
```

**Q20. Write a Python function to flatten a nested list.**

```python
def flattened_list(nested_list)-> list:
    flat_list = []
    for num in nested_list:
        if isinstance(num,list):
            flat_list.extend(flattened_list(num))
        else:
            flat_list.append(num)
    return flat_list
nested_list = [[1, 2, 3], 4, [6, 7]]
print(flattened_list(nested_list))
```

```python
def binary_Search(arr,target)-> int:
    low, high = 0,len(arr)-1
    
    while low <= high:
        mid = (low + high) // 2
        print(arr[mid])

        if arr[mid] < target:
            low = mid + 1
        elif arr[mid] > target:
            high = mid - 1
        else:
            return mid

    return -1
arr = [1,58,63,74,84,96,99,101,120,310,560]
target = int(input("Enter a value to find: "))
print(binary_Search(arr,target))
```

**Q23. How do you find the longest common prefix in a list of strings?**

```python
def common_prefix(strr)->str:
    if not str:
        return "there is no string value in the list"
    prefix = strr[0]
    for s in strr[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
    return prefix
words = ["flower","flow","flew"]
print(common_prefix(words))

```

**Q24. How do you use Python’s match-case statement introduced in modern Python versions?**

**Q25. How do you implement a stack using a list?**

```python
class Stack:
    def __init__(self):
        self.stack = []
    def push(self,item):
        self.stack.append(item)
    def pop(self):
        return self.stack.pop() if not self.is_empty() else None
    def is_empty(self):
        return len(self.stack)==0

s= Stack()
s.push(10)
s.push(20)
s.push(30)
s.push(40)
print(s.stack)
print(s.pop())
print(s.push(30))
print(s.stack)
print(s.is_empty())
```

**Q27. How do you rotate a list to the right by k steps?**

```python
def list_rotation(lst,k)-> list:
    k = k % len(lst)
    return lst[-k:] + lst[:-k]
lst= [3,4,5,6,7,8]
k = 7
print(list_rotation(lst,k))
```

**Q28. Write a function to check if a given string is a valid palindrome considering only alphanumeric characters.**

```python
def palindrome_Sentence(sent)-> str:
    pharse =""
    for s in sent.lower():
        if s.isalnum():
            pharse += s
    return pharse == pharse[::-1]
sent = "A man, a plan, a canal: Panama"
print(palindrome_Sentence(sent)) 
            
```

**Q29. How do you calculate the power of a number using recursion?**

```python
def power(base,exp):
    if exp == 0:
        return 1
    else:
        return base * (power(base,exp-1))

print(power(2,3))
```

#### BANK SYSTEM FOR WITHDRAWAL AND DEPOSIT:

```python
## Create a class BankAccount with deposit() and withdraw() methods, and an attribute balance. Prevent withdrawal if balance is insufficient.
class BankAccount:
    def __init__(self,balance=0):
        self.balance=balance
    def deposit(self,amount):
        self.balance += amount
        print(f"The updated balance amount is {self.balance}")
    def withdraw(self,amount):
        if amount > self.balance:
            print("insufficient balance")
        else:
            self.balance -= amount
            print(f"the updated balance after withdrawal is {self.balance}")
acc = BankAccount(10000)
acc.deposit(1000)
acc.withdraw(500)
```
